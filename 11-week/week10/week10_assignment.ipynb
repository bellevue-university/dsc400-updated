# Part 1: Lecture - Introduction to AI Models for Big Data Analytics

Introduction to AI Models for Big Data Analytics:

In big data analytics, AI techniques play a crucial role in extracting valuable insights from large and complex datasets. This lecture introduces two commonly used AI models for big data analytics: clustering and anomaly detection.

Key Concepts:

1. Clustering:
   - Clustering is the process of grouping similar data points together based on their characteristics. It helps in identifying patterns and structures within the data.
   - KMeans is a popular clustering algorithm that partitions data into 'k' clusters by minimizing the within-cluster variance.

2. Anomaly Detection:
   - Anomaly detection is the identification of observations that deviate significantly from the norm in a dataset. It is useful for detecting unusual patterns or outliers.
   - Isolation Forest is an effective algorithm for anomaly detection, particularly in high-dimensional datasets, by isolating anomalies in the feature space.

Python Libraries:

- scikit-learn: A powerful library in Python for machine learning tasks, including clustering and anomaly detection.

Applications:

- Clustering: Useful for customer segmentation, recommendation systems, and image segmentation.
- Anomaly Detection: Applied in fraud detection, network security, and equipment monitoring.

```python
# Import necessary libraries
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer, TextDataset, DataCollatorForLanguageModeling, Trainer, TrainingArguments
```

```python
import os
os.environ["WANDB_DISABLED"] = "true"
```

```python
# Load pre-trained GPT-2 model and tokenizer
model_name = 'gpt2-medium'  # Change to the desired pre-trained model size
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)
```

```python
# Load and preprocess your dataset
train_dataset = TextDataset(
    tokenizer=tokenizer,
    file_path="train.txt",  # Path to your training dataset file
    block_size=128  # Adjust block size as needed
)
```

```python
# Define training arguments and trainer
training_args = TrainingArguments(
    output_dir="llm-fine-tuned",     # Directory to save the fine-tuned model
    overwrite_output_dir=True,
    num_train_epochs=3,              # Number of training epochs
    per_device_train_batch_size=8,   # Batch size during training
    save_steps=1000,                 # Save model checkpoint every 1000 steps
    save_total_limit=2,              # Limit the number of saved checkpoints
    prediction_loss_only=True,       # Only compute the prediction loss
    report_to="none",                # <--- completely disables WandB / MLflow / tensorboard logging
    logging_steps=50                 # Still logs to stdout
)
```

```python
# Initialize trainer
trainer = Trainer(
    model=model,
    args=training_args,
    data_collator=DataCollatorForLanguageModeling(
        tokenizer=tokenizer, mlm=False
    ),
    train_dataset=train_dataset,
)
```

```python
# Fine-tune the pre-trained model on your dataset
trainer.train()
```

```python
# Save the fine-tuned model
model.save_pretrained("llm-fine-tuned")
```

```python
# Load fine-tuned model for testing
fine_tuned_model = GPT2LMHeadModel.from_pretrained("llm-fine-tuned")

# Define a prompt or input text for generation
prompt = "In recent years, artificial intelligence has revolutionized"

# Tokenize the prompt
input_ids = tokenizer.encode(prompt, return_tensors="pt")

# Generate text based on the prompt
output = fine_tuned_model.generate(input_ids, max_length=100, num_return_sequences=1, temperature=0.7, do_sample=True)

# Decode and print the generated text
generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print("Generated Text:")
print(generated_text)
```
